```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")
print(df.shape)
print(df.head())
df.info()
print(df.isnull().sum())
print(df.describe())

target = "survived"
print(df[target].value_counts())
print(df[target].value_counts(normalize= True))
print(df.groupby("sex")[target].mean())          # survival rate per sex
```
